# Study 01: OpenFDA MAUDE Z-Score Sensitivity & ClinicalTrials.gov Landscape — Insulin Infusion Pumps

This notebook reproduces the empirical analysis from `01_insulin_pump_maude_zscore_study.md` using zero-cost public FDA & NIH APIs (`2,056,124` MAUDE MDRs analyzed on `2026-11-12 UTC`).


```python
import json
from pathlib import Path

data = json.loads(Path('artifacts/insulin_pump_live_dataset.json').read_text(encoding='utf-8'))
ms = data['maude_summary']
print('Total MAUDE events:', f"{ms['total_events_analyzed']:,}")
print('Severity breakdown:', ms['event_type_counts'])
print('Default 12-month spikes (Z>=2.0):', ms['signal_spikes'])
print('Extended 35-month spikes (Z>=2.0):', data['extended_analysis_36m']['z_threshold_2_0_spikes'])
```

```python
# Inspect top device problem categories
for prob in ms['top_device_problems']:
    print(f"{prob['problem']:55s} | {prob['count']:8,d} ({prob['pct']:.1f}%)")
```

### Methodological Takeaway
1. A default 12-month trailing slice (`signal_spikes = []`) absorbed late-2025 peaks inside its burn-in window.
2. Extending to 35 complete months flagged two July reporting volume excursions (`2024-07`: `Z = +5.11`; `2025-07`: `Z = +2.10`) **for review** (not causal safety findings — MAUDE counts reflect batch submissions/reporting lags, never incidence rates).
3. Final EU MDR / MDCG 2022-21 safety conclusions require PRRC review alongside internal sales-unit denominators.
